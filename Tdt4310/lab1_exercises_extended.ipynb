## Questions:

**1. A language model (LM) is...**

A machine learning model trained to use probability to arrange words and sequences in a sentence.

**2. A large language model (LLM) is...**

A deep learning algorithm that can process and generate complex content with a lot of parameters by using massive datasets.

**3. A pre-trained base/foundation model is...**

Large neural networks trained on massive amounts of data, which imbues the model with a profound understanding of grammar, context and semantics.

**4. What do we mean by fine-tuned language models, and what does the *head* mean in this context?**

A fine-tuned language model is a model given specialized training based on spesific datasets or domains for spesific tasks.
The "head" refers to the additional layers used when fine-tuning to adapt the language understanding capabilities of the LLM for spesific task requirements.

**5. Which one is more suitable for a chatbot: a pretrained model or a pretrained + fine-tuned model?**


A pretrained and fine-tuned chatbot is better for both general purpose chatbots and industrial utilization, where more in depth understanding of context might be necessary, as it is needed to understand the nuances of the spesific tasks (like question-answering or reviewing, chatGPT was fine-tuned by reinforced learning from human feedback (RLHF)).

**6. What do we mean by knowledge cutoff in models like ChatGPT?**

The term "knowledge cutoff" refers to the point when the pre-training data stops being collected. Once this phase ends, access to news, information or new data needs to be supplemented.

**7. What do we mean when we talk about confabulations/hallucinations of LLMs?**


These are scenarios where the LLM intertwine genuine and false data or creates entirelly fictious facts to try and present it as factual.

**8. The context length (e.g. 4096) of a GPT-based model is...**


The maximum number of tokens the model can remember when generating text, they matter as the tokens have an effect on the models comprehension, allowing it to give more suitable answers

**9. Why is it problematic to increase the context/sequence length for transformer-based models?**


It comes back to processing power, as it demands more computation and memory complexity, which can result in inefficient processing.

**10. Parameters are used to describe the model size of LLMs (such as 7B, 13B, …). What do these parameters represent, and what makes LLMs difficult to interpret?**


These parameters represent adjustable parameters in a model: the model size (complexity and processing), training data (perfomance of the model) and hyperparameters (prefomance settings for training). Thanks to the complexity of these models with billions of parameters it becomes difficult to interpret why they generate particular outputs, and to measure its reproducibility and consistency.

**11. Quantization (e.g. through techniques like LoRa, GPTQ, AWQ) is typically used to reduce the size of LLMs allow them to be run on consumer-grade hardware - even laptops! Explain the main idea behind quantization:**


Quantizaion is a method of compressing LLMs by converting the weights (matrix of numerical values) into more efficient data types, increasing its efficiency and optimization while sacrifising some accuracy.

**12. LLMs need vast training data. The size of training data is typically referred to as tokens. What is a token and what happens to unknown words outside of the models' vocabulary?**


Tokens are chunks of text and can be characters, words or strings read or generated based on context. Any new word that wasn't in the training gets replaced by the most likely replacement based on neighboring words and context.

**13. What is the difference between model training and model inference?**


In the model training phase all of the data it is fed it will learn and analyze from, while in the interference phase, it will make predictions from interpreted data.

**14.  Why are GPUs (graphics processing units) ideal for training larger transformer-based models?**


GPUs have a large number of cores which allow for better computation and multiple parallel processes giving it support towards complex, multi-step processes, as the training requires inputting large continous data sets to expand and redefine what the algorithm can do.

**15.  What is the biggest technical limitation with respect to GPUs for training LLMs?**


While GPUs offer high parallel processing power, they often have limited memory and demand more power for processing.

**16. LLMs aimed towards human interactions should be *aligned*. What does this mean?**


LLM Alignment is about making sure that the models follow human preferences, language patterns and values.
