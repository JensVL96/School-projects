```python
from typing import List

# this is the base class of tokenizers in nltk
from nltk.tokenize.api import TokenizerI
from nltk.tokenize import WordPunctTokenizer, TweetTokenizer, ToktokTokenizer


# this is just a simple example of how a tokenizer can be implemented
class MyWhitespaceTokenizer(TokenizerI):
    def __init__(self):
        super().__init__()

    def tokenize(self, text: str) -> List[str]:
        return text.split()


sentence = "That U.S.A. poster-print costs $12.40..."

# ************************************************************
# TODO: import and add the tokenizers you want to try out here
# ************************************************************
tokenizers = [
    MyWhitespaceTokenizer(),
    WordPunctTokenizer(),
    TweetTokenizer(),
    ToktokTokenizer()
]

# Leave this function as-is
def tokenize(tokenizers: List[TokenizerI], sentence: str) -> None:
    for tokenizer in tokenizers:
        assert isinstance(tokenizer, TokenizerI)
        tokenized = tokenizer.tokenize(sentence)
        print(f"{tokenizer.__class__.__name__} ({len(tokenized)} tokens)\n{tokenized}\n")


tokenize(tokenizers, sentence)
```

MyWhitespaceTokenizer (5 tokens):  
    ['That', 'U.S.A.', 'poster-print', 'costs', '$12.40...']  
    This tokenizer recognises each part of the sentence as it retains the logic of the country USA, the hyphen poster-print and the price as one single token.

WordPunctTokenizer (16 tokens):  
    ['That', 'U', '.', 'S', '.', 'A', '.', 'poster', '-', 'print', 'costs', '$', '12', '.', '40', '...']  
    This tokenizer splits the sentence into the most amount of tokens as it fails to recognise any connections.

TweetTokenizer (12 tokens):  
    ['That', 'U', '.', 'S', '.', 'A', '.', 'poster-print', 'costs', '$', '12.40', '...']  
    This tokenizer recognizes some special symbols, like the connecting hyphen and the dot in the price, however it didn't retain USA or that the price was in dollars.

ToktokTokenizer (5 tokens):  
    ['That', 'U.S.A.', 'poster-print', 'costs', '$', '12.40', '...']  
    This tokenizer is able to recognize most of the same tokens as the (white)space tokenizer, except for missing the currency symbols to the price.

# 2. Language modeling
We have now studied the bigger models like BERT and GPT-based language models. A simpler language model, however, can implemented using n-grams.

**1. What is an n-gram?**

An n-gram is a contiguous sequence of words or characters extracted from a given text or speech sample, providing insight into local patterns and dependencies within the language.

**2. Use NLTK to print out bigrams and trigrams for the given sentence below. Your function should support any number of N.**

```python
import nltk
from nltk import word_tokenize
from nltk.util import ngrams
nltk.download('punkt')

sentence = "That U.S.A. poster-print costs $12.40... I'd pay $5.00 for it."

def generate_ngrams(sentence, N):
    # Tokenize the sentence into words
    words = word_tokenize(sentence)

    # Generate n-grams using NLTK's ngrams function
    n_grams = ngrams(words, N)

    # Return the n-grams
    return n_grams

# Generate and print bigrams (2-grams)
res_bigrams = generate_ngrams(sentence, 2)
for gram in res_bigrams:
    print(f"bigram: {gram}")

# Generate and print trigrams (3-grams)
res_trigrams = generate_ngrams(sentence, 3)
for gram in res_trigrams:
    print(f"trigram: {gram}")
```

**3. Based on your intuition for language modeling, how can n-grams be used for word predictions?**

By estimating the probability of the next word based on the context of the previous n-1 words.

**4. NLTK includes the `FreqDist` class, which produces the frequency distribution of words in a sentence. Use it to print out the two most common words in the text below.**

```python
from nltk import FreqDist
from nltk.tokenize import word_tokenize

# Given text
text = "That that is is that that is not. Is that it? It is. You sure? Surely it is!"

# Tokenize the text into words
words = word_tokenize(text)

# Calculate the frequency distribution
freq_dist = FreqDist(words)

# Print the two most common words
most_common_words = freq_dist.most_common(2)
print("Two most common words:", most_common_words)
```

**5. Use your n-gram function from question 2.2 to print out the most common trigram of the text in question 2.4**

```python
from nltk import FreqDist
# Given text
text = "That that is is that that is not. Is that it? It is. You sure? Surely it is!"

lowercase = text.lower()

# Generate trigrams
trigrams = generate_ngrams(lowercase, 3)

# Calculate the frequency distribution of trigrams
trigram_freq_dist = FreqDist(trigrams)

# Print the most common trigram
most_common_trigram = trigram_freq_dist.most_common(1)
print("Most common trigram:", most_common_trigram)
```

**1. Describe the main differences between bag-of-words and one-hot encoding through examples.**

Bag-of-words represents text as a frequency-based vector, capturing word counts, while one-hot encoding uses binary vectors to represent the presence or absence of individual words.

**2. What are the limitations of the above representations?**

The limitations of bag-of-words and one-hot encoding include the loss of semantic meaning and the inability to capture word relationships or context.

**3. Example of word embedding techniques, such as Word2Vec and GloVe are considered *dense* representations. How do dense word embeddings relate to the *distributional hypothesis*?**

By representing words as continuous vectors in a dense vector space, where the proximity of vectors reflects semantic similarity, this aligns with the idea that words with similar meanings tend to occur together in context.
