## Questions:

**1. A language model (LM) is...**

A machine learning model trained to use probability to arrange words and sequences in a sentence.

**2. A large language model (LLM) is...**

A deep learning algorithm capable of processing and generating complex content with numerous parameters by using massive datasets.

**3. A pre-trained base/foundation model is...**

Large neural networks trained on extensive datasets, giving the model with a great understanding of grammar, context, and semantics.

**4. What do we mean by fine-tuned language models, and what does the *head* mean in this context?**

A fine-tuned language model undergoes specialized training on specific datasets or domains for spesific tasks. The "head" refers to additional layers used during fine-tuning to adapt the language understanding capabilities of the LLM for specific task requirements.

**5. Which one is more suitable for a chatbot: a pretrained model or a pretrained + fine-tuned model?**


A pretrained and fine-tuned chatbot is better for both general-purpose chatbots and industrial utilization, where a more in-depth understanding of context might be necessary, such as question-answering or reviewing. ChatGPT, for example, was fine-tuned using reinforced learning from human feedback (RLHF).

**6. What do we mean by knowledge cutoff in models like ChatGPT?**

The term "knowledge cutoff" refers to the point when the pre-training data stops being collected. Once this phase ends, access to news, information, or new data needs to be supplemented.

**7. What do we mean when we talk about confabulations/hallucinations of LLMs?**


These are scenarios where LLMs intertwine genuine and false data or create entirely fictitious facts, attempting to present them as factual.

**8. The context length (e.g. 4096) of a GPT-based model is...**


The maximum number of tokens the model can remember when generating text. Context length matters as tokens affect the model's comprehension, enabling it to provide more suitable answers.

**9. Why is it problematic to increase the context/sequence length for transformer-based models?**


Increasing the context/sequence length demands more computational and memory complexity, leading to inefficient processing.

**10. Parameters are used to describe the model size of LLMs (such as 7B, 13B, …). What do these parameters represent, and what makes LLMs difficult to interpret?**


These parameters represent adjustable parameters in a model: the model size, training data, and hyperparameters. Due to the complexity of these models with billions of parameters, it becomes difficult to interpret why they generate particular outputs and to measure their reproducibility and consistency.

**11. Quantization (e.g. through techniques like LoRa, GPTQ, AWQ) is typically used to reduce the size of LLMs allow them to be run on consumer-grade hardware - even laptops! Explain the main idea behind quantization:**


Quantization is a method of compressing LLMs by converting weights (a matrix of numerical values) into more efficient data types, increasing efficiency and optimization while sacrificing some accuracy.

**12. LLMs need vast training data. The size of training data is typically referred to as tokens. What is a token and what happens to unknown words outside of the models' vocabulary?**


Tokens are chunks of text, such as characters, words, or strings, read or generated based on context. Any new word not in the training data gets replaced by the most likely replacement based on neighboring words and context.

**13. What is the difference between model training and model inference?**


In the model training phase, the model learns and analyzes from all the data it is fed. In the inference phase, it makes predictions from interpreted data.

**14.  Why are GPUs (graphics processing units) ideal for training larger transformer-based models?**


GPUs have a large number of cores, allowing for better computation and multiple parallel processes that support complex, multi-step processes during training, which involves inputting large continuous datasets to expand and redefine the algorithm's capabilities.

**15.  What is the biggest technical limitation with respect to GPUs for training LLMs?**


While GPUs offer high parallel processing power, they often have limited memory and demand more power for processing.

**16. LLMs aimed towards human interactions should be *aligned*. What does this mean?**


LLM Alignment involves ensuring that models follow human preferences, language patterns, and values.

**17. RLHF (reinforcement learning from human feedback) is used in models like ChatGPT. While you are not expected to learn about reinforcement learning (RL), explain the following terms in context of LLMs (1 sentence each):**

🔺 this question might be too hard without any knowledge of RL. feel free to skip it. 🔺

- SFT (supervised fine-tuning)
    - Updating a pre-trained language model using labeled data for a specific task while manually supervising the entire training phase.
- Reward model
    - A large language model designated for sending a ranking signal to the original language model in training, optimizing its parameters.
- Policy model
    - A model that evolves by learning the actions it needs to take to lead to positive rewards.

**18. For the above procedures, we require data, ideally labeled by humans. This is called data annotation. In this context, what is inter-annotator agreement, and why is it important for model creation?**


Inter-annotator agreement compares the annotations (labeling data to provide context) of multiple annotators. It is important for quality assurance, improving the accuracy and overall performance of the model.

**19. Do you believe LLM projects should be open-sourced? Why?**


Yes, I believe LLM projects should be open-sourced, as transparency bolsters trust in the model, and providing access gives more control and flexibility to adapt and potentially innovate the model.

**20. Why is it important to consider the ethical implications of LLMs?**

Some ethical implications of LLMs include the risk of misuse in factual sources like academic papers, biased outputs from only one perspective, and breaches of privacy as they can gather information without referencing its sources.

**21. *Smaller* language models ($\leq 100M$ parameters) will likely be important in the days to come. What are some motivations for smaller models?**

In general, smaller LMs require fewer computational resources for training and operation, making them more affordable, with faster inference times and ease of tailoring to tasks or applications where resources are limited.
