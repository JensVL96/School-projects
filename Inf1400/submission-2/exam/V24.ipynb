```python
# * En bygning er en generell driftsbygning som finnes på en gård. Den har følgende egenskaper: 
class Bygning:
    def __init__(self, maks_plasser):
        # § Et antall plasser til dyr (maksimalt antall)  
        self.maks_plasser = maks_plasser
        # § En liste som lagrer på dyrene i bygningen  
        self.dyreliste = []

    # § Den tilbyr funksjonalitet for å sette inn nye dyr i bygningen  
    def settInn(self, art):
        if len(self.dyreliste) >= self.maks_plasser:
            raise ValueError("Ingen flere plasser tilgjengelig i bygningen")
        self.dyreliste.append(art)

    # § Den tilbyr funksjonalitet for å hente ut antall dyr som står i bygningen 
    def hvorMange(self):
        return len(self.dyreliste)
    
    # § En __str__-metode som henter en string som sier hvor mange dyr som står i 
    # bygningen, og som skriver ut denne informasjonen ved bruk av print.
    def __str__(self):
        return f"Bygning med {self.hvorMange()} dyr."
    
    # * Det skal også være mulig å selge dyr ved å bruke metoden «selg». Denne tar et antall
    # dyr som skal selges som parameter, og fjerner dette antallet fra bygningens liste over
    # dyr. Om det ikke finnes så mange dyr i bygningen som man ønsker å selge, skal det
    # raises en ValueError.
    def selg(self, salgsantall):
        # Unødvendig duplikat: if len(self.dyreliste) < salgsantall:
        if self.hvorMange() < salgsantall:
            raise ValueError("Det er ikke så mange dyr i bygningen")

        for _ in range(salgsantall):
            self.dyreliste.pop()

class Hønsehus(Bygning):
    def __init__(self, maks_plasser):
        super().__init__(maks_plasser)
        # self.høner = 0
        # self.haner = 0

    def settInn(self, art):
        if not isinstance(art, Hønsefugl):
            raise TypeError("Kun hønsefugler kan settes inn i hønsehuset")
        super().settInn(art)
        if art.kjønn == "hane":
            self.haner += 1
        elif art.kjønn == "høne":
            self.høner += 1
        
    # def __str__(self):
    #     return f"Hønsehus med {self.haner} haner og {self.høner} høner."
    
    #Eksempel fra eksamensløsning:
    # def __str__(self):
    #     høne_teller = 0
    #     hane_teller = 0
    #     for fugl in self.dyreliste:
    #         if fugl.kjønn == "høne":
    #             høne_teller += 1
    #         else:
    #             hane_teller += 1
    #     return f"Hønsehus med {hane_teller} haner og {høne_teller} høner."
    
    # Egen vri på ny løsning:
    #Eksempel fra eksamensløsning:
    def __str__(self):
        høne_teller = sum(1 for dyr in self.dyreliste if dyr.kjønn == "høne")
        hane_teller = sum(1 for dyr in self.dyreliste if dyr.kjønn == "hane")
        return f"Hønsehus med {hane_teller} haner og {høne_teller} høner."

#* Et hønsehus er en bygning som kun kan inneholde høns. Den skal ha de samme
#egenskapene som klassen Bygning, men ved innsetting av dyr skal det garanteres for
#at dyret som setter inn er av typen Hønsefugl. __str__-metoden skal ikke hente ut
#antall dyr, men returnere en string som forteller hvor mange høner og hvor mange
#haner det er i hønsehuset.
#* Eksempelutskrift for __str__: «Hønsehus med 2 haner og 10 høner»

class Fjøs(Bygning):
    def __init__(self, antall):
        super().__init__(antall)

    def settInn(self, art):
        if not isinstance(art, Ku):
            raise TypeError("Kun kyr kan settes inn i fjøset")
        super().settInn(art)

    def __str__(self):
        return f"Fjøs med {self.hvorMange()} kyr."
#* Et fjøs er en bygning for kyr. Den har de samme egenskapene som klassen Bygning,
#men det er kun lov til å sette inn objekter av klassen Ku i et fjøs. Ettersom det ikke
#holdes okser på gårder som brukes systemet vil det ikke være nødvendig å skille på
#ku og okse i __str__-metoden, men den skal endres slik at man kan se at det er et
#fjøs.
#* Eksempelutskrift for __str__: «Fjøs med 8 kyr»
```

```python
class Hønsefugl:
    def __init__(self, kjønn):
        self.kjønn = kjønn
        if kjønn not in ["høne", "hane"]:
            raise ValueError("Kjønn må være enten hane eller høne")

    def legg_egg(self):
        if self.kjønn == "høne":
            return 2
        return 0


class Ku:
    def __init__(self):
        self.melkeproduksjon = 20

    def melk(self):
        mengde = self.melkeproduksjon
        self.melkeproduksjon = 0
        return mengde
```

```python
class Bondegård:
    def __init__(self):
        # Inneholder en datastruktur som holder på bygninger på gården. Du velger datastruktur selv.
        self.bygninger = {}
        # Inneholder en variabel som holder på antall dyr som finnes på gården til enhver tid.
        self.antall_dyr = 0

    # Metode for å legge til en ny bygning på gården
    def leggTilBygning(self, navn, bygning):
        self.bygninger[navn] = bygning

    # Tilbyr en metode for å sette inn et nytt dyr i en bygning. Metoden tar bygningens navn og et
    # dyre-objekt som parametere. Om bygningen ikke finnes, eller det ikke er plass til dyret i
    # bygningen, raises en ValueError.
    def settInnDyr(self, navn, dyr):
        if navn not in self.bygninger.keys():
            raise ValueError("Bygningen eksisterer ikke.")
        
        # settInn metoden reiser allerede en error hvis det ikke er plass til dyret, så det er ikke nødvendig med enda en verdisjekk her
        bygning = self.bygninger[navn]
        bygning.settInn(dyr)
        self.antall_dyr += 1

    # Tilbyr en metode for å selge dyr fra gården. Parameterne er navnet på bygningen det skal
    # selges fra, og antallet dyr som skal selges. Det skal raises en passende error dersom
    # bygningen ikke eksisterer, eller om det ikke finnes nok dyr i bygningen til å selge.
    def selge(self, navn, antall):
        if navn not in self.bygninger:
            raise ValueError("Bygningen eksisterer ikke.")
        
        # selg metoden reiser allerede en error om det ikke finnes nok dyr i bygningen til å selge.
        bygning = self.bygninger[navn]
        bygning.selg(antall)
        self.antall_dyr -= 1

    # Tilbyr en metode for å hente egg. Metoden løper gjennom datastrukturen du har valgt og
    # finner alle hønsehus. I hvert hønsehus hentes det egg fra alle høner. Returnerer antallet egg
    # som ble hentet.
    def hent_egg(self):
        antall_egg = 0
        for bygning in self.bygninger.values():
            if isinstance(bygning, Hønsehus):
                for dyr in bygning.dyreliste:
                    antall_egg += dyr.legg_egg()
        return antall_egg

    # Tilbyr en metode for å melke alle kyr. Metoden løper gjennom datastrukturen du har valgt og
    # finner alle fjøs. Alle kyr i fjøsene melkes, og det returneres hvor mange liter melk man har
    # melket.
    def melk_kyr(self):
        liter_melk = 0
        for bygning in self.bygninger.values():
            if isinstance(bygning, Fjøs):
                for dyr in bygning.dyreliste:
                    liter_melk += dyr.melk()
        return liter_melk

    # Tilbyr en __str__-metode som lar oss printe ut informasjon om alle bygninger på gården.
    def __str__(self):
        result = "Bondegården har følgende bygninger:\n"
        for navn, bygning in self.bygninger.items():
            result += f"{navn}: {bygning}\n"
        return result


if __name__ == "__main__":
    b = Bondegård()
    b.bygninger["Fjøs1"] = Fjøs(10)
    b.bygninger["Hønsehus1"] = Hønsehus(5)

    b.settInnDyr("Fjøs1", Ku())
    b.settInnDyr("Fjøs1", Ku())
    b.settInnDyr("Hønsehus1", Hønsefugl("høne"))

    print(b)

    print(b.hent_egg())
    print(b.melk_kyr())

```
