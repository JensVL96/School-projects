Både betjent hytte og Selvbetjent hytte klassen benytter samme metoder som hytte klassen, så det gir mening at de arver fra denne. Siden de to underklassene benytter seg av samme metoder, men med ulike "edge" tilfeller, gir et også mening å benytte polymorfi, da metoden kan luke ut eventuelle tilfeller, for så å bruke den overholdende metoden.

Oppgave 1C
Skriv klassene, komplett med metoder og instansvariabler

```python
class Hytte:
    def __init__(self, navn, pris, kapasitet):
        self.hyttenavn = navn
        self.pris = pris
        self.kapasitet = kapasitet
        self.innsjekket = []

    def innsjekk(self, person):
        if len(self.innsjekket) >= self.kapasitet:
            raise RuntimeError("Det er ikke plass på hytta")
        self.innsjekket.append(person)

    def finnPerson(self, navn):
        return any(gjest.navn == navn for gjest in self.innsjekket)

    def __str__(self):
        return f"{self.hyttenavn}; antall innsjekkede: {len(self.innsjekket)}"


class BetjentHytte(Hytte):
    def __init__(self, navn, pris, kapasitet):
        super().__init__(navn, pris, kapasitet)
        self.verter = []

    def innsjekk(self, person):
        if isinstance(person, Vert):
            self.verter.append(person)
        else:
            super().innsjekk(person)

    def finnPerson(self, navn):
        return any(person.navn == navn for person in self.innsjekket + self.verter)

    def __str__(self):
        return f"{super().__str__()}, Betjent hytte"


class SelvbetjentHytte(Hytte):
    def __init__(self, navn, pris, kapasitet):
        super().__init__(navn, pris, kapasitet)

    def innsjekk(self, person):
        if isinstance(person, Vert):
            raise TypeError("Selvbetjente hytter har ingen verter.")
        else:
            super().innsjekk(person)

    def __str__(self):
        return f"{super().__str__()}, Selvbetjent hytte"


class Person:
    def __init__(self, navn):
        self.navn = navn


class Vert(Person):
    pass


class Gjest(Person):
    pass
```

Ved å gi en tekststreng som parameter sender du kun den vesentlige informasjonen metoden trengerog den vil også være enkel å bruke fra flere klasser uten navnekonflikter. Ev fordel ved å sende klasseinstans istedet, hadde vært at kilden på parameteren hadde vært enkel å spore, men da som nevnt tidligere fungerer dette bedre hvis metoden bare blir brukt fra samme kilden.


vi vet at alle hyttene benytter samme metoder, og inneholder navn som instansevariabel. Så finnperson kan bli benyttet med en enkel streng, derfor trenger ikke metoden å motta klasse
det er positivt å bruke string som parametertype for metoden fordi at alle hytter allerede har metoder som finner en person basert på string. Derfor kan vi bare bruke den metoden fra alle hyttene (så slipper vi å implementere noe nytt).

Oppgave 3B
Skriv klassen Hyttesystem med alle tilhørende instansvariabler og metoder.

```python
class Hyttesystem:
    def __init__(self):
        self.hytter = []
        self.gjester = []
        self.verter = []

    def leggTilHytte(self, hyttenavn, pris, kapasitet, erBetjent):
        if erBetjent:
            self.hytter.append(BetjentHytte(hyttenavn, pris, kapasitet))
        else:
            self.hytter.append(SelvbetjentHytte(hyttenavn, pris, kapasitet))

    def leggTilGjest(self, navn):
        self.gjester.append(Gjest(navn))

    def leggTilVert(self, navn, lønn):
        self.verter.append(Vert(navn, lønn))

    def hvorEr(self, navn):
        for hytte in self.hytter:
            if hytte.finnPerson(navn):
                return hytte.hyttenavn
        return "Finner ikke person"
    
    def __str__(self):
        result = "Hytter i systemet:\n"
        for hytte in self.hytter:
            result += str(hytte) + "\n"
        return result
```

```python
# Oppretter et Hyttesystem-objekt
system = Hyttesystem()

# Legger til noen hytter
system.leggTilHytte("Hytte 1", 500, 6, True)
system.leggTilHytte("Hytte 2", 300, 4, False)
system.leggTilHytte("Hytte 3", 600, 8, True)

# Legger til noen gjester
system.leggTilGjest("Gjest 1")
system.leggTilGjest("Gjest 2")
system.leggTilGjest("Gjest 3")

# Legger til noen hytteverter
system.leggTilVert("Vert 1", 20000)
system.leggTilVert("Vert 2", 25000)
system.leggTilVert("Vert 3", 18000)

# Får gjestene til å sjekke inn på hytter
system.gjester[0].sjekkInn(system.hytter[0])
system.gjester[1].sjekkInn(system.hytter[1])
system.gjester[2].sjekkInn(system.hytter[2])

# Får hytteverter til å begynne å jobbe på hytter
system.verter[0].fåArbeid(system.hytter[0])
system.verter[1].fåArbeid(system.hytter[1])
system.verter[2].fåArbeid(system.hytter[2])

# Får gjestene til å sjekke inn på hytter
system.hytter[0].innsjekk(system.gjester[0])
system.hytter[1].innsjekk(system.gjester[1])
system.hytter[2].innsjekk(system.gjester[2])

# Får hytteverter til å begynne å jobbe på hytter
system.hytter[0].innsjekk(system.verter[0])
# system.hytter[1].innsjekk(system.verter[1])
system.hytter[2].innsjekk(system.verter[2])



# Skriver ut Hyttesystem-objektet
print(system)
```
